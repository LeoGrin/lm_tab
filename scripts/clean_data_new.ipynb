```python
from auto_mm_bench.datasets import dataset_registry
import pandas as pd

print(dataset_registry.list_keys())  # list of all dataset names
dataset_name = 'product_sentiment_machine_hack'

train_dataset = dataset_registry.create(dataset_name, 'train')
test_dataset = dataset_registry.create(dataset_name, 'test')

# Concatenate train and test datasets
combined_dataset = pd.concat([train_dataset.data, test_dataset.data], ignore_index=True)

to_remove = ["Unnamed: 0", "Text_ID"]
target_name = "Sentiment"

# Remove unnecessary columns
combined_dataset = combined_dataset.drop(columns=to_remove)

# Check for missing values
missing_values = combined_dataset.isnull().sum()
print("Missing values in each column:")
print(missing_values)

# Drop rows with missing target values
combined_dataset = combined_dataset.dropna(subset=[target_name])

# rename target name to target
combined_dataset = combined_dataset.rename(columns={target_name: "target"})

combined_dataset.to_parquet("data/product_sentiment_machine_hack.parquet")
```
